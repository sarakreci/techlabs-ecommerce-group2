```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Faylın adını bura yaz (əgər Excel-dirsə pd.read_excel('faylin_adi.xlsx') yazmalısan)
df = pd.read_csv('online_retail_II.csv', encoding='latin1')

# Datanın ilk 5 sətrinə baxaq
print(df.head())

# Sütun adlarını görək ki, 'Price' və ya digər adların necə yazıldığını bilək
print(df.columns)

# Rəqəmsal sütunların statistikası
print(df.describe())
```

```python
# Qiymət (Price) sütununun paylanması
plt.figure(figsize=(10, 5))
sns.histplot(df['Price'], bins=30, kde=True, color='blue')
plt.title('Distribution of Price')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()

# Outlier-ləri görmək üçün boxplot
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['Price'], color='orange')
plt.title('Boxplot of Price')
plt.show()
```
